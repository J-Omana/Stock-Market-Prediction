```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler(feature_range=(0,1))
scaled_data = scaler.fit_transform(Closing_price[['Close']].values)
scaled_data
```

```python

scaled_data.shape
```

```python
x_data = []
y_data = []

base_days = 100

for i in range(base_days,len(scaled_data)):
    x_data.append(scaled_data[i-base_days:i])
    y_data.append(scaled_data[i])

x_data, y_data = np.array(x_data), np.array(y_data)
```

```python
x_data[0], y_data[0]
```

```python

len_train = int(len(x_data)*0.9)
x_train = x_data[:len_train]
y_train = y_data[:len_train]

x_test = x_data[len_train:]
y_test = y_data[len_train:]
```

```python

print(x_train.shape)
print(y_train.shape)
print(x_test.shape)
print(y_test.shape)
```

```python
from keras.models import Sequential
from keras.layers import Dense, LSTM, Input

model = Sequential()

# Define the input shape using the Input layer
model.add(Input(shape=(x_train.shape[1], 1)))

model.add(LSTM(128, return_sequences=True))
model.add(LSTM(64, return_sequences=False))
model.add(Dense(25))
model.add(Dense(1))
```

```python
model.summary()
```

```python
model.compile(optimizer='adam', loss='mean_squared_error')
```

```python
model.fit(x_train, y_train, batch_size = 5, epochs=10)
```

```python

predictions = model.predict(x_test)
predictions
```

```python

inv_predictions = scaler.inverse_transform(predictions)
inv_predictions
```

```python
inv_y_test = scaler.inverse_transform(y_test)
```

```python
import numpy as np

def predict_future(no_of_days, prev_100):
    future_predictions = []
    for i in range(no_of_days):
        # Ensure prev_100 is a numpy array with the correct shape
        prev_100_array = np.array(prev_100).reshape(1, -1, 1)
        
        # Predict the next day
        next_day = model.predict(prev_100_array)
        
        # Update prev_100 by sliding the window
        prev_100 = prev_100_array.tolist()  # Ensure compatibility with list operations
        prev_100[0].append(next_day[0])  # Append the predicted value
        prev_100[0] = prev_100[0][1:]      # Remove the first element
        
        # Add the inverse transformed prediction to future_predictions
        inverse_transformed = scaler.inverse_transform([[next_day[0][0]]])
        future_predictions.append(inverse_transformed[0][0])
    
    return future_predictions
```

```python
# Prepare last_100 for prediction
last_100 = Closing_price[['Close']].tail(100)
last_100 = scaler.transform(last_100.values.reshape(-1, 1)).reshape(1, -1, 1)
last_100 = last_100.tolist()  # Convert to list for predict_future function

# Predict the next 100 days
no_of_days = 10
future_results = predict_future(no_of_days, last_100)

# Print results
print(future_results)
```

```python

future_results = np.array(future_results).reshape(-1,1)
plt.figure()
pd.DataFrame(future_results).plot(figsize=(15,5), marker='o')
for i in range(len(future_results)):
    plt.text(i,future_results[i],int(future_results[i][0]))
plt.xlabel('Future days')
plt.ylabel('Close price')
plt.title("Future Close price of Bit coin")
```
